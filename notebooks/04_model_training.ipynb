```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

X_train = pd.read_csv("../agent/dataset/data/X_train.csv")
X_test = pd.read_csv("../agent/dataset/data/X_test.csv")
y_train = pd.read_csv("../agent/dataset/data/y_train.csv").squeeze()
y_test = pd.read_csv("../agent/dataset/data/y_test.csv").squeeze()

cheat_columns = ["transport_cost_inr", "hotel_cost_inr", "food_cost_inr", "activity_cost_inr", "other_cost_inr", "cost_per_km", "daily_spending"]
X_train = X_train.drop(columns=cheat_columns)
X_test = X_test.drop(columns=cheat_columns)

baseline_model = LinearRegression()

baseline_model.fit(X_train, y_train)

predictions = baseline_model.predict(X_test)

mae = mean_absolute_error(y_test, predictions)
r2 = r2_score(y_test, predictions)

print("=== BASELINE MODEL RESULTS (NO CHEATING) ===")
print(f"Mean Absolute Error (MAE): ₹{mae:.2f}")
print(f"Accuracy Score (R2): {r2 * 100:.2f}%")
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(n_estimators=100, random_state=42)

print("Training the Random Forest (this might take a few seconds)...")
rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)

rf_mae = mean_absolute_error(y_test, rf_predictions)
rf_r2 = r2_score(y_test, rf_predictions)

print("\n=== ADVANCED MODEL RESULTS (RANDOM FOREST) ===")
print(f"Mean Absolute Error (MAE): ₹{rf_mae:.2f}")
print(f"Accuracy Score (R2): {rf_r2 * 100:.2f}%")
print(f"\nImprovement over Baseline: ₹{mae - rf_mae:.2f} less error per prediction!")
```
