```python
import pandas as pd
import requests
import time

df = pd.read_csv("../agent/dataset/data/travelsense_india_trips.csv")
unique_destinations = df["destination"].unique()

print(f"Found {len(unique_destinations)} unique destinations to scrape!\n")

scraped_data = []

for dest in unique_destinations:
    print(f"Fetching data for {dest}...")
    
    url = f"https://en.wikipedia.org/api/rest_v1/page/summary/{dest}"
    
    try:
        # We add a User-Agent header because Wikipedia requires it
        headers = {"User-Agent": "TravelSenseAI/1.0 (Student Project)"}
        response = requests.get(url, headers=headers)
        data = response.json()
        
        description = data.get("extract", "No description available.")
        image_url = ""
        
        if "thumbnail" in data:
            image_url = data["thumbnail"]["source"]
            
        scraped_data.append({
            "name": dest,
            "description": description,
            "image_url": image_url
        })
        
    except Exception as e:
        print(f"Failed to fetch {dest}: {e}")
        
    time.sleep(1)
    
destinations_df = pd.DataFrame(scraped_data)
destinations_df.to_csv("../agent/dataset/data/scraped_destinations.csv", index=False)

print("\n=== SCRAPING COMPLETE ===")
display(destinations_df.head())
```
