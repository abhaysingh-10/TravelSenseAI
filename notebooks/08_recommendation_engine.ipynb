```
import pandas as pd

# 1. Load the data
trips_df = pd.read_csv("../agent/dataset/data/travelsense_india_trips.csv")
destinations_db = pd.read_csv("../agent/dataset/data/scraped_destinations.csv")

# 2. Calculate the "vibe" of each destination
dest_stats = trips_df.groupby("destination").agg(
    avg_cost=("total_cost_inr", "mean"),
    avg_days=("trip_days", "mean")
).reset_index()

# 3. Merge stats with Wikipedia data
master_destinations = pd.merge(dest_stats, destinations_db, left_on="destination", right_on="name")

# 4. Recommendation Engine Logic
def get_recommendations(traveler_type):
    print(f"\n=== AI PICKS FOR: {traveler_type.upper()} ===")
    
    if traveler_type == "Luxury Travelers":
        recs = master_destinations.sort_values(by="avg_cost", ascending=False).head(3)
    elif traveler_type == "Frequent Backpackers":
        recs = master_destinations.sort_values(by="avg_cost", ascending=True).head(3)
    else:
        recs = master_destinations[(master_destinations["avg_cost"] > 30000) & (master_destinations["avg_cost"] < 50000)].head(3)
        
    for index, row in recs.iterrows():
        name = row["name"]
        cost = row["avg_cost"]
        img = row["image_url"]
        print(f"🌍 {name}")
        print(f"💰 Avg Cost: ₹{cost:,.0f}")
        print(f"📸 Image: {img}\n")

# 5. Let test our engine!
get_recommendations("Luxury Travelers")
get_recommendations("Frequent Backpackers")
```
