```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

X_train_cost = pd.read_csv("../agent/dataset/data/X_train.csv")
X_test_cost = pd.read_csv("../agent/dataset/data/X_test.csv")
y_train_cost = pd.read_csv("../agent/dataset/data/y_train.csv").squeeze()
y_test_cost = pd.read_csv("../agent/dataset/data/y_test.csv").squeeze()

y_train_duration = X_train_cost["trip_days"]
y_test_duration = X_test_cost["trip_days"]

X_train_duration = X_train_cost.drop(columns=["trip_days"])
X_train_duration["total_cost_inr"] = y_train_cost

X_test_duration = X_test_cost.drop(columns=["trip_days"])
X_test_duration["total_cost_inr"] = y_test_cost

cheat_columns = ["transport_cost_inr", "hotel_cost_inr", "food_cost_inr", "activity_cost_inr", "other_cost_inr", "cost_per_km", "daily_spending"]
X_train_duration = X_train_duration.drop(columns=cheat_columns)
X_test_duration = X_test_duration.drop(columns=cheat_columns)

baseline_dur_model = LinearRegression()
baseline_dur_model.fit(X_train_duration, y_train_duration)

dur_predictions = baseline_dur_model.predict(X_test_duration)
mae_dur = mean_absolute_error(y_test_duration, dur_predictions)
r2_dur = r2_score(y_test_duration, dur_predictions)

print("=== BASELINE MODEL RESULTS (DURATION PREDICTION) ===")
print(f"Mean Absolute Error (MAE): {mae_dur:.2f} Days")
print(f"Accuracy Score (R2): {r2_dur * 100:.2f}%")
```
